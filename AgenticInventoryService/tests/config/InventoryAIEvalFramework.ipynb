# New questions and answers plus other useful attributes for testing and regression defense

new set of attributes and explanations:
* test/eval name:  ex. "verify_resort_typo": 
* description: example: "Check if a resort exists with a possible typo.",
* category": "inventory_text", (use case category, ex. inventory, demand prediction, advanced analysis/comparison)
* user_request": "Is Tusckanny Villoge a resort name?",
* expected_answer": like "Did you mean Tuscany Village, a Hilton Grand Vacations Club?",
* difficulty": "easy", (t-shirts size for now)
* exact_response": False, (eval instructions (exact for data pulls to table or data in text)
* is_edge_case": True, (cases designed to test functionality and prove accuracy for tricky questions and reduce unexpected results)
* is_test_case": True, (canon test suite--run for LLM-side of CI/CD or key commits)
* is_graphical": False,(text/table vs. text for eval)
* expected_agent_flow": "A1, A2,"

```python
# Convert q_a_dict to a Pandas DataFrame
df = pd.DataFrame.from_dict(q_a_dict, orient='index')


# Function to query based on criteria
def query_tests(category=None, difficulty=None, graphical=None, edge_case=None):
    """Filter test cases based on criteria."""
    filtered_df = df.copy()

    if category:
        filtered_df = filtered_df[filtered_df["category"] == category]
    if difficulty:
        filtered_df = filtered_df[filtered_df["difficulty"] == difficulty]
    if graphical is not None:
        filtered_df = filtered_df[filtered_df["is_graphical"] == graphical]
    if edge_case is not None:
        filtered_df = filtered_df[filtered_df["is_edge_case"] == edge_case]

    return filtered_df
```
